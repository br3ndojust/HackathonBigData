```python
import pandas as pd
import numpy as np
```

```python
produtos = pd.DataFrame({
    "produto_id": [f"P{i:03}" for i in range(1, 51)],  # 50 produtos
    "categoria": np.random.choice(["Cerveja", "Refrigerante", "Água", "Destilado"], 50),
    "descricao": [f"Produto {i}" for i in range(1, 51)],
    "caracteristica_1": np.random.choice(["Premium", "Popular", "Edição Especial"], 50),
    "caracteristica_2": np.random.choice(["Lata", "Garrafa", "Long Neck"], 50),
    "caracteristica_3": np.random.choice(["350ml", "500ml", "1L"], 50),
    "caracteristica_4": np.random.choice(["Importado", "Nacional"], 50)
})
```

```python
pdvs = pd.DataFrame({
    "pdv_id": [f"S{i:03}" for i in range(1, 21)],  # 20 PDVs
    "on_off_prem": np.random.choice(["On-Prem", "Off-Prem"], 20),
    "categoria_pdv": np.random.choice(["c-store", "g-store", "liquor", "bar"], 20),
    "zipcode": np.random.randint(10000, 99999, 20)
})
```

```python
n_transacoes = 10000
datas = pd.date_range("2022-01-01", "2022-12-31", freq="D")

transacoes = pd.DataFrame({
    "data_transacao": np.random.choice(datas, n_transacoes),
    "pdv_id": np.random.choice(pdvs["pdv_id"], n_transacoes),
    "produto_id": np.random.choice(produtos["produto_id"], n_transacoes),
    "quantidade": np.random.randint(1, 20, n_transacoes)
})

# Preço médio por produto para faturamento
precos = {p: np.random.uniform(2, 50) for p in produtos["produto_id"]}
transacoes["faturamento"] = transacoes.apply(
    lambda x: x["quantidade"] * precos[x["produto_id"]], axis=1
)
```
